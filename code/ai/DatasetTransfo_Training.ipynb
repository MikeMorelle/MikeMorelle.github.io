```python
import os
import shutil
import random
from pathlib import Path

dataset_A = Path("/kaggle/input/home-fire-dataset")
dataset_B = Path("/kaggle/input/indoor-fire-and-smoke-detection-with-yolov8/data")
output = Path("/content/fire_dataset")

TRAIN_SIZE = 1000
VAL_SIZE = 200
SEED = 42

random.seed(SEED)

if output.exists():
    shutil.rmtree(output)

# out dir
for split in ["train", "val"]:
    (output / "images" / split).mkdir(parents=True, exist_ok=True)
    (output / "labels" / split).mkdir(parents=True, exist_ok=True)

samples = []

image_extensions = [".jpg", ".jpeg", ".png", ".bmp", ".webp"]

def collect_dataset(images_dir, labels_dir, prefix):

    if not images_dir.exists():
        print(f"WARNING: no imgs: {images_dir}")
        return

    if not labels_dir.exists():
        print(f"WARNING: no lbls: {labels_dir}")
        return

    for image_path in images_dir.iterdir():

        if image_path.suffix.lower() not in image_extensions:
            continue

        label_path = labels_dir / f"{image_path.stem}.txt"

        if not label_path.exists():
            print(f"WARNING: no label for {image_path.name}")
            continue

        lines = label_path.read_text().splitlines()

        fire_lines = []

        for line in lines:
            parts = line.split()

            if len(parts) >= 5 and parts[0] == "0":
                fire_lines.append(line)

        #  only take fire
        if not fire_lines:
            continue

        samples.append({
            "image": image_path,
            "label": label_path,
            "name": f"{prefix}_{image_path.stem}",
            "extension": image_path.suffix
        })

# ============================================================
# DATASET A
# train + valid + test
# ============================================================

for split in ["train", "val", "test"]:

    collect_dataset(
        dataset_A / split / "images",
        dataset_A / split / "labels",
        prefix=f"A_{split}"
    )

    if split=="val":
      split = "valid"

    collect_dataset(
        dataset_B / split / "images",
        dataset_B / split / "labels",
        prefix=f"B_{split}"
    )

print("\n================================")
print("Collected Samples")
print("================================")

print(f"Gesamt: {len(samples)}")

random.shuffle(samples)

if len(samples) < TRAIN_SIZE + VAL_SIZE:
    raise ValueError(
        f"only found: {len(samples)}, "
        f"need: {TRAIN_SIZE + VAL_SIZE}"
    )

selected_samples = samples[:TRAIN_SIZE + VAL_SIZE]

train_samples = selected_samples[:TRAIN_SIZE]
val_samples = selected_samples[TRAIN_SIZE:TRAIN_SIZE + VAL_SIZE]


def copy_samples(samples, split):

    for sample in samples:

        image_name = sample["name"] + sample["extension"]
        label_name = sample["name"] + ".txt"

        shutil.copy2(
            sample["image"],
            output / "images" / split / image_name
        )

        original_lines = sample["label"].read_text().splitlines()

        fire_lines = [
            line for line in original_lines
            if len(line.split()) >= 5 and line.split()[0] == "0"
        ]

        with open(output / "labels" / split / label_name, "w") as f:
            f.write("\n".join(fire_lines))


copy_samples(train_samples, "train")
copy_samples(val_samples, "val")


# ============================================================
# create data.yaml
# ============================================================

yaml_content = f"""path: {output}

train: images/train
val: images/val

nc: 1

names:
  0: Fire
"""

with open(output / "data.yaml", "w") as f:
    f.write(yaml_content)



print("\n================================")
print("MERGE COMPLETED")
print("================================")

print(f"Train: {len(train_samples)}")
print(f"Val:   {len(val_samples)}")
print(f"Total: {len(samples)}")

print("\nStructure:")

for path in sorted(output.rglob("*")):
    print(path)
```

```python
import shutil
from pathlib import Path

knife_dataset = Path("/content/knife_dataset")
fire_dataset = Path("/content/fire_dataset")

output = Path("/content/knife_fire_dataset")

if output.exists():
    shutil.rmtree(output)


for split in ["train", "val"]:
    (output / "images" / split).mkdir(parents=True, exist_ok=True)
    (output / "labels" / split).mkdir(parents=True, exist_ok=True)



def merge_dataset(dataset_path, prefix, class_mapping):

    for split in ["train", "val"]:

        images_dir = dataset_path / "images" / split
        labels_dir = dataset_path / "labels" / split

        if not images_dir.exists():
            print(f"WARNING: {images_dir} not found")
            continue

        if not labels_dir.exists():
            print(f"WARNING: {labels_dir} not found")
            continue

        for image_path in images_dir.iterdir():

            if image_path.suffix.lower() not in [
                ".jpg", ".jpeg", ".png",
                ".bmp", ".webp"
            ]:
                continue

            label_path = labels_dir / f"{image_path.stem}.txt"

            if not label_path.exists():
                continue

            new_image_name = f"{prefix}_{image_path.name}"
            new_label_name = f"{prefix}_{image_path.stem}.txt"

            shutil.copy2(
                image_path,
                output / "images" / split / new_image_name
            )

            new_labels = []

            for line in label_path.read_text().splitlines():

                parts = line.split()

                if len(parts) < 5:
                    continue

                old_class = int(parts[0])

                if old_class in class_mapping:

                    new_class = class_mapping[old_class]

                    parts[0] = str(new_class)

                    new_labels.append(" ".join(parts))

            with open(
                output / "labels" / split / new_label_name,
                "w"
            ) as f:

                f.write("\n".join(new_labels))

# ============================================================
# KNIFE DATASET
#
# 0 = Knife
# ============================================================

merge_dataset(
    knife_dataset,
    prefix="knife",
    class_mapping={
        0: 0
    }
)


# ============================================================
# FIRE DATASET
#
# 0 = Fire
#
# becomes:
#
# 1 = Fire
# ============================================================

merge_dataset(
    fire_dataset,
    prefix="fire",
    class_mapping={
        0: 1
    }
)


# ============================================================
# data.yaml
# ============================================================

yaml_content = f"""path: {output}

train: images/train
val: images/val

nc: 2

names:
  0: Knife
  1: Fire
"""

with open(output / "data.yaml", "w") as f:
    f.write(yaml_content)

print("\n================================")
print("MERGE COMPLETED")
print("================================")

for split in ["train", "val"]:

    images = list(
        (output / "images" / split).glob("*")
    )

    labels = list(
        (output / "labels" / split).glob("*.txt")
    )

    print(
        f"{split:5} | "
        f"Images: {len(images):5} | "
        f"Labels: {len(labels):5}"
    )

print("\nOutput:")
print(output)

print("\ndata.yaml:")
print((output / "data.yaml").read_text())
```

```python
from pathlib import Path
import shutil

src = Path("/content/own_dataset")
dst = Path("/content/own_filtered_dataset")

splits = ["train", "valid"]

for split in splits:
  img_src = src / split / "images"
  lbl_src = src / split / "labels"

  img_dst = dst / "images" / ("train" if split=="train" else "val")
  lbl_dst = dst / "labels" / ("train" if split=="train" else "val")

  img_dst.mkdir(parents=True, exist_ok=True)
  lbl_dst.mkdir(parents=True, exist_ok=True)

  for label_file in lbl_src.glob("*.txt"):
    new_lines  = []

    with open(label_file) as f:
      for line in f:
        parts = line.strip().split()
        cls = int(parts[0])

        if cls in keep:
          parts[0] = str(keep[cls])
          new_lines.append(" ".join(parts))

    if new_lines:
      out_label = lbl_dst / label_file.name
      with open(out_label, "w") as f:
        f.write("\n".join(new_lines))

      stem = label_file.stem
      for ext in [".jpg", ".jpeg", ".png"]:
        img = img_src / (stem + ext)
        if img.exists():
          shutil.copy2(img, img_dst / img.name)
          break
```

```python
import os
import shutil
from pathlib import Path

SOURCE_DATASET = Path("/content/knife_dataset")
TARGET_DATASET = Path("/content/knife_fire_dataset/content/knife_fire_dataset")

# class in knife_dataset
SOURCE_KNIFE_CLASS = 3

# knife in our dataset
TARGET_KNIFE_CLASS = 0


def copy_knife_samples(source_split, target_split):
    source_images = SOURCE_DATASET / source_split / "images"
    source_labels = SOURCE_DATASET / source_split / "labels"

    target_images = TARGET_DATASET / "images" / target_split
    target_labels = TARGET_DATASET / "labels" / target_split

    target_images.mkdir(parents=True, exist_ok=True)
    target_labels.mkdir(parents=True, exist_ok=True)

    image_extensions = [".jpg", ".jpeg", ".png", ".bmp", ".webp"]

    copied = 0

    for image_path in source_images.iterdir():

        if image_path.suffix.lower() not in image_extensions:
            continue

        label_path = source_labels / f"{image_path.stem}.txt"

        if not label_path.exists():
            continue

        knife_annotations = []

        with open(label_path, "r") as f:
            for line in f:
                parts = line.strip().split()

                if len(parts) < 5:
                    continue

                class_id = int(parts[0])

                # only keep knife
                if class_id == SOURCE_KNIFE_CLASS:
                    # class 3 -> class 0
                    parts[0] = str(TARGET_KNIFE_CLASS)
                    knife_annotations.append(" ".join(parts))

        if not knife_annotations:
            continue

        #prefix e
        new_stem = f"e_{source_split}_{image_path.stem}"

        target_image = target_images / f"{new_stem}{image_path.suffix}"
        target_label = target_labels / f"{new_stem}.txt"

        shutil.copy2(image_path, target_image)


        with open(target_label, "w") as f:
            f.write("\n".join(knife_annotations) + "\n")

        copied += 1

    print(
        f"{source_split} -> {target_split}: "
        f"{copied} Knife-Samples kopiert"
    )

copy_knife_samples("train", "train")


print("Finished.")
```

```python
from collections import Counter
from pathlib import Path

TARGET_DATASET = Path("/content/knife_fire_dataset/content/knife_fire_dataset")

CLASS_NAMES = {
    0: "Knife",
    1: "Fire"
}


def analyze_split(split):
    labels_dir = TARGET_DATASET / "labels" / split
    images_dir = TARGET_DATASET / "images" / split

    class_counter = Counter()
    image_count = 0
    annotation_count = 0

    image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

    for image_path in images_dir.iterdir():
        if image_path.suffix.lower() in image_extensions:
            image_count += 1

    for label_path in labels_dir.glob("*.txt"):

        with open(label_path, "r") as f:
            for line in f:
                parts = line.strip().split()

                if len(parts) < 5:
                    continue

                class_id = int(parts[0])

                class_counter[class_id] += 1
                annotation_count += 1

    print(f"\n{'=' * 40}")
    print(f"{split.upper()}")
    print(f"{'=' * 40}")

    print(f"Images:       {image_count}")
    print(f"Annotations: {annotation_count}")

    for class_id, class_name in CLASS_NAMES.items():
        count = class_counter[class_id]

        percentage = (
            count / annotation_count * 100
            if annotation_count > 0
            else 0
        )

        print(
            f"{class_name:10s}: "
            f"{count:6d} "
            f"({percentage:6.2f} %)"
        )

analyze_split("train")
analyze_split("val")
```

```python
from ultralytics import YOLO

model = YOLO("yolo11n.pt")

model.train(
    data="/content/knife_fire_dataset/content/knife_fire_dataset/data.yaml",

    epochs=100,
    imgsz=640,

    # ========================================================
    # Lighting / Color
    # ========================================================

    hsv_h=0.015,
    hsv_s=0.45,
    hsv_v=0.25,

    # ========================================================
    # Geometric Augmentation
    # ========================================================

    # Especially important for knives at different angles
    degrees=25,

    # Vary the object's position in the image
    translate=0.10,

    # different object sizes
    scale=0.35,

    # slight change in perspective
    shear=3.0,
    perspective=0.0005,

    # ========================================================
    # Reflection
    # ========================================================

    fliplr=0.5,
    flipud=0.0,

    # ========================================================
    # Multi-object / Different Image Compositions
    # ========================================================

    mosaic=0.5,
    mixup=0.0,

    # ========================================================
    # Training
    # ========================================================

    batch=-1,
    device=-1,
    patience=10,
    verbose=True
)
```
