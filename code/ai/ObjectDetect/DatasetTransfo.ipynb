```python
import fiftyone as fo
import fiftyone.zoo as foz
import fiftyone.types as fot
from fiftyone import ViewField as F
```

Open images with Axe, Baseball bat, Chainsaw, Drill, Glove, Screwdriver, Hammer, Weapon, Knife, Scissors

```python
labels = ["Hammer", "Baseball bat", "Glove", "Knife", "Scissors"]

train_dataset = foz.load_zoo_dataset(
        "open-images-v7",
        split="train",
        label_types=["detections"],
        classes=labels,
        max_samples = 5000,
)

val_dataset = foz.load_zoo_dataset(
        "open-images-v7",
        split="validation",
        label_types=["detections"],
        classes=labels,
        max_samples = 500,
)
```

Filter only relevant labels from the dataset

```python
train_dataset = train_dataset.filter_labels(
    "ground_truth",
    F("label").is_in(labels)
)

val_dataset = val_dataset.filter_labels(
    "ground_truth",
    F("label").is_in(labels)
)
```

How often does each label occur?

```python
counts = train_dataset.count_values("ground_truth.detections.label")

for c in labels:
    print(f"{c}: {counts.get(c, 0)}")

print("______________________________")

counts = val_dataset.count_values("ground_truth.detections.label")

for c in labels:
    print(f"{c}: {counts.get(c, 0)}")
```

```python
keep = {
    0: 0, # fire
    1: 1, # knife
    3: 2, # scissors
}
```

```python
from pathlib import Path
import shutil

src = Path("/content/own_dataset")
dst = Path("/content/own_filtered_dataset")

splits = ["train", "valid"]

for split in splits:
  img_src = src / split / "images"
  lbl_src = src / split / "labels"

  img_dst = dst / "images" / ("train" if split=="train" else "val")
  lbl_dst = dst / "labels" / ("train" if split=="train" else "val")

  img_dst.mkdir(parents=True, exist_ok=True)
  lbl_dst.mkdir(parents=True, exist_ok=True)

  for label_file in lbl_src.glob("*.txt"):
    new_lines  = []

    with open(label_file) as f:
      for line in f:
        parts = line.strip().split()
        cls = int(parts[0])

        if cls in keep:
          parts[0] = str(keep[cls])
          new_lines.append(" ".join(parts))

    if new_lines:
      out_label = lbl_dst / label_file.name
      with open(out_label, "w") as f:
        f.write("\n".join(new_lines))

      stem = label_file.stem
      for ext in [".jpg", ".jpeg", ".png"]:
        img = img_src / (stem + ext)
        if img.exists():
          shutil.copy2(img, img_dst / img.name)
          break
```

##Using fire dataset from kaggle

0 = Feuer, 1 = Smoke

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("pengbo00/home-fire-dataset")

print("Path to dataset files:", path)
```

```python
from collections import defaultdict
import os

label_dir = os.path.join(path, "train", "labels")

img_to_classes = {}

for filename in os.listdir(label_dir):
    if not filename.endswith(".txt"):
        continue

    with open(os.path.join(label_dir, filename), "r") as f:
      classes = {int(line.split()[0]) for line in f}

    img_name = filename.replace(".txt", ".jpg")
    img_to_classes[img_name] = classes
```

```python
candidates = []

for img, classes in img_to_classes.items():
  if 0 in classes or 1 in classes:
    candidates.append(img)

print(len(candidates))
```

```python
import random

random.shuffle(candidates)

train_subset = candidates[:1000]
val_subset = candidates[1000:1100]

print(len(train_subset))
print(len(val_subset))
```

```python
import shutil

img_src = os.path.join(path, "train", "images")
label_src = os.path.join(path, "train", "labels")

splits = {
    "train": train_subset,
    "val": val_subset
}

for split, ds in splits.items():
  out_img = f"/content/fire_dataset/images/{split}"
  out_label = f"/content/fire_dataset/labels/{split}"

  os.makedirs(out_img, exist_ok=True)
  os.makedirs(out_label, exist_ok=True)

  for img in ds:
    img_path = os.path.join(img_src, img)
    label_path = os.path.join(label_src, img.replace(".jpg", ".txt"))

    if os.path.exists(img_path) and os.path.exists(label_path):
      shutil.copy(img_path, out_img)
      shutil.copy(label_path, out_label)

print(len(os.listdir(out_img)))
print(len(os.listdir(out_label)))
```

```python
mapping_openimages = {
    0:0,  # hammer
    1:1,  # baseball bat
    2:2,  # glove
    3:3,  # knife
    4:4   # scissors
}

mapping_own = {
    0:5,  # fire
    1:3,  # knife
    2:4   # scissors
}

mapping_fire = {
    0:5,  # fire
    1:6   # smoke
}
```

Since the naming conventions differ completely across the three datasets, there is no need to address duplicate filenames or file overwriting.

```python
from pathlib import Path
import shutil

def merge_dataset(src, dst, mapping):
  src = Path(src)
  dst = Path(dst)

  for split in ["train", "val"]:
    (dst/"images"/split).mkdir(parents=True, exist_ok=True)
    (dst/"labels"/split).mkdir(parents=True, exist_ok=True)

    img_src = src/"images"/split
    lbl_src = src/"labels"/split

    for lbl in lbl_src.glob("*.txt"):
      new_lines = []

      with open(lbl) as f:
        for line in f:
          parts = line.strip().split()
          cls = int(parts[0])

          if cls in mapping:
            parts[0] = str(mapping[cls])
            new_lines.append(" ".join(parts))

      if len(new_lines)==0:
        continue

      with open(dst/"labels"/split/lbl.name, "w") as f:
        f.write("\n".join(new_lines))

      stem=lbl.stem

      for ext in [".jpg", ".png", ".jpeg"]:
        img = img_src/(stem+ext)

        if img.exists():
          shutil.copy2(img, dst/"images"/split/img.name)
          break
```

```python
merge_dataset("/content/dataset", "/content/merged_dataset", mapping_openimages)
merge_dataset("/content/own_filtered_dataset", "/content/merged_dataset", mapping_own)
merge_dataset("/content/fire_dataset", "/content/merged_dataset", mapping_fire)
```

```python
import yaml

data = {
    "path": "/content/merged_dataset",
    "train": "images/train",
    "val": "images/val",
    "names": {
        0: "Hammer",
        1: "Baseball bat",
        2: "Glove",
        3: "Knife",
        4: "Scissors",
        5: "Fire",
        6: "Smoke"
    }
}

with open("/content/merged_dataset/data.yaml", "w") as f:
  yaml.dump(data, f, sort_keys=False)
```
